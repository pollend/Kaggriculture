## 2. Official 1.32.2 evaluator


```python
def score_from_state(env, seat):
    final = env.steps[-1][seat]
    try:
        reward = float(final.reward)
    except Exception:
        reward = 0.0

    if reward != 0:
        return reward

    for states in reversed(env.steps):
        try:
            return float(
                states[0].observation.farms[seat].money
            )
        except Exception:
            continue

    return reward


def leftover_value(env, seat):
    final_state = env.steps[-1][seat]
    observation = final_state.observation
    private = getattr(observation, "private", {}) or {}
    shed = private.get("shed", {}) or {}
    inventories = private.get("inventories", []) or []
    prices = (
        getattr(observation, "market", {}) or {}
    ).get("prices", {}) or {}

    total = 0.0
    for item, quantity in shed.items():
        total += float(prices.get(item, 0) or 0) * float(quantity or 0)

    for inventory in inventories:
        for item, quantity in (inventory or {}).items():
            total += float(prices.get(item, 0) or 0) * float(quantity or 0)

    return total


def play(left_name, right_name, seed):
    env = make(
        "kaggriculture",
        configuration={
            "episodeSteps": 720,
            "seed": int(seed),
        },
        debug=False,
    )
    env.run([
        str(PATHS[left_name]),
        str(PATHS[right_name]),
    ])
    final = env.steps[-1]

    return {
        "left_money": score_from_state(env, 0),
        "right_money": score_from_state(env, 1),
        "left_leftover": leftover_value(env, 0),
        "right_leftover": leftover_value(env, 1),
        "left_status": str(final[0].status),
        "right_status": str(final[1].status),
    }


def symmetric_pair(candidate, opponent, seed):
    first = play(candidate, opponent, seed)
    second = play(opponent, candidate, seed)

    rows = [
        {
            "candidate": candidate,
            "opponent": opponent,
            "seed": seed,
            "seat": 0,
            "candidate_money": first["left_money"],
            "opponent_money": first["right_money"],
            "candidate_leftover": first["left_leftover"],
            "opponent_leftover": first["right_leftover"],
            "candidate_status": first["left_status"],
            "opponent_status": first["right_status"],
        },
        {
            "candidate": candidate,
            "opponent": opponent,
            "seed": seed,
            "seat": 1,
            "candidate_money": second["right_money"],
            "opponent_money": second["left_money"],
            "candidate_leftover": second["right_leftover"],
            "opponent_leftover": second["left_leftover"],
            "candidate_status": second["right_status"],
            "opponent_status": second["left_status"],
        },
    ]

    for row in rows:
        row["margin"] = (
            row["candidate_money"]
            - row["opponent_money"]
        )
        row["result"] = (
            "win"
            if row["margin"] > 0
            else "loss"
            if row["margin"] < 0
            else "tie"
        )

    return rows


EVAL_AVAILABLE = ENGINE_READY and bool(CONTROL_SOURCES)

if EVAL_AVAILABLE:
    first_control = next(iter(CONTROL_SOURCES))
    smoke = play(
        "Anchor Exact",
        first_control,
        27001,
    )
    EVAL_AVAILABLE = (
        smoke["left_status"] == "DONE"
        and smoke["right_status"] == "DONE"
        and not (
            smoke["left_money"] <= 1
            and smoke["right_money"] <= 1
        )
    )
    print({
        "smoke": smoke,
        "evaluation_available": EVAL_AVAILABLE,
    })
else:
    print({
        "evaluation_available": False,
        "reason": ENGINE_ERROR,
    })
```

## 3. Stage A — direct ablation against the exact baseline


```python
ANCHOR_NAME = "Anchor Exact"
STAGE_A_SEEDS = (27011, 27031)

if EVAL_AVAILABLE:
    stage_a_rows = []

    for candidate in CANDIDATE_SOURCES:
        if candidate == ANCHOR_NAME:
            continue

        for seed in STAGE_A_SEEDS:
            pair = symmetric_pair(
                candidate,
                ANCHOR_NAME,
                seed,
            )
            pair_margin = sum(
                row["margin"]
                for row in pair
            )

            for row in pair:
                row["pair_margin"] = pair_margin
                stage_a_rows.append(row)

    stage_a_summary = []

    for candidate in sorted({
        row["candidate"]
        for row in stage_a_rows
    }):
        group = [
            row
            for row in stage_a_rows
            if row["candidate"] == candidate
        ]
        margins = [row["margin"] for row in group]
        monies = [
            row["candidate_money"]
            for row in group
        ]
        leftovers = [
            row["candidate_leftover"]
            for row in group
        ]
        wins = sum(
            row["result"] == "win"
            for row in group
        )
        losses = sum(
            row["result"] == "loss"
            for row in group
        )
        pair_margins = {
            row["seed"]: row["pair_margin"]
            for row in group
        }

        stage_a_summary.append({
            "candidate": candidate,
            "games": len(group),
            "wins": wins,
            "ties": len(group) - wins - losses,
            "losses": losses,
            "net_wins": wins - losses,
            "positive_pairs": sum(
                value > 0
                for value in pair_margins.values()
            ),
            "mean_money": round(
                statistics.mean(monies),
                2,
            ),
            "minimum_money": min(monies),
            "mean_margin": round(
                statistics.mean(margins),
                2,
            ),
            "minimum_margin": min(margins),
            "mean_leftover_value": round(
                statistics.mean(leftovers),
                2,
            ),
            "selection_score": round(
                (wins - losses) * 1_000_000
                + sum(
                    value > 0
                    for value in pair_margins.values()
                ) * 100_000
                + statistics.mean(margins)
                - 0.01 * statistics.mean(leftovers),
                2,
            ),
        })

    stage_a_summary.sort(
        key=lambda row: row["selection_score"],
        reverse=True,
    )
    STAGE_A_FINALISTS = [
        row["candidate"]
        for row in stage_a_summary[:2]
    ]
    print(json.dumps(stage_a_summary, indent=2))
else:
    stage_a_rows = []
    stage_a_summary = []
    STAGE_A_FINALISTS = []
    print({
        "stage_a": "SKIPPED",
        "reason": "Reliable 1.32.2 evaluator unavailable.",
    })
```

## 4. Stage B — broad money and leftover-inventory gate


```python
CONTROL_NAMES = tuple(CONTROL_SOURCES)
STAGE_B_SEEDS = (27101,)

if EVAL_AVAILABLE:
    tested = [ANCHOR_NAME, *STAGE_A_FINALISTS]
    stage_b_rows = []

    for candidate in tested:
        for opponent in CONTROL_NAMES:
            for seed in STAGE_B_SEEDS:
                stage_b_rows.extend(
                    symmetric_pair(
                        candidate,
                        opponent,
                        seed,
                    )
                )

    stage_b_summary = []

    for candidate in tested:
        group = [
            row
            for row in stage_b_rows
            if row["candidate"] == candidate
        ]
        margins = [row["margin"] for row in group]
        monies = [
            row["candidate_money"]
            for row in group
        ]
        leftovers = [
            row["candidate_leftover"]
            for row in group
        ]
        wins = sum(
            row["result"] == "win"
            for row in group
        )
        losses = sum(
            row["result"] == "loss"
            for row in group
        )

        direct = next(
            (
                row
                for row in stage_a_summary
                if row["candidate"] == candidate
            ),
            {
                "net_wins": 0,
                "positive_pairs": 0,
                "mean_margin": 0,
            },
        )

        stage_b_summary.append({
            "candidate": candidate,
            "games": len(group),
            "wins": wins,
            "ties": len(group) - wins - losses,
            "losses": losses,
            "mean_money": round(
                statistics.mean(monies),
                2,
            ),
            "minimum_money": min(monies),
            "mean_margin": round(
                statistics.mean(margins),
                2,
            ),
            "minimum_margin": min(margins),
            "mean_leftover_value": round(
                statistics.mean(leftovers),
                2,
            ),
            "direct_net_wins": direct["net_wins"],
            "direct_positive_pairs": direct["positive_pairs"],
            "direct_mean_margin": direct["mean_margin"],
        })

    print(json.dumps(stage_b_summary, indent=2))
else:
    stage_b_rows = []
    stage_b_summary = []
```

## 5. Conservative promotion


```python
def find_summary(rows, name):
    return next(
        (
            row
            for row in rows
            if row["candidate"] == name
        ),
        None,
    )


if EVAL_AVAILABLE and STAGE_A_FINALISTS:
    anchor_broad = find_summary(
        stage_b_summary,
        ANCHOR_NAME,
    )
    eligible = []

    for name in STAGE_A_FINALISTS:
        direct = find_summary(
            stage_a_summary,
            name,
        )
        broad = find_summary(
            stage_b_summary,
            name,
        )

        direct_ok = (
            direct is not None
            and direct["net_wins"] > 0
            and direct["positive_pairs"] >= 1
            and direct["mean_margin"] > 0
            and direct["mean_money"] > 0
        )

        broad_ok = (
            broad is not None
            and anchor_broad is not None
            and broad["wins"] >= anchor_broad["wins"]
            and broad["mean_money"]
                > anchor_broad["mean_money"]
            and broad["minimum_money"]
                >= anchor_broad["minimum_money"] * 0.99
            and broad["mean_leftover_value"]
                <= anchor_broad["mean_leftover_value"] * 1.02
        )

        if direct_ok and broad_ok:
            score = (
                direct["net_wins"] * 1_000_000
                + direct["positive_pairs"] * 100_000
                + direct["mean_margin"]
                + broad["mean_money"]
                - 0.01 * broad["mean_leftover_value"]
            )
            eligible.append((score, name))

    if eligible:
        eligible.sort(reverse=True)
        SELECTED_NAME = eligible[0][1]
    else:
        SELECTED_NAME = ANCHOR_NAME
else:
    SELECTED_NAME = ANCHOR_NAME

SELECTED_SOURCE = CANDIDATE_SOURCES[SELECTED_NAME]

print({
    "selected": SELECTED_NAME,
    "fallback_to_anchor": (
        SELECTED_NAME == ANCHOR_NAME
    ),
})
```

## 6. Package the verified winner


```python
MAIN_PATH.write_text(
    SELECTED_SOURCE,
    encoding="utf-8",
)
ast.parse(SELECTED_SOURCE)
compile(
    SELECTED_SOURCE,
    str(MAIN_PATH),
    "exec",
)

raw = io.BytesIO()
with gzip.GzipFile(
    fileobj=raw,
    mode="wb",
    filename="",
    mtime=0,
) as zipped:
    with tarfile.open(
        fileobj=zipped,
        mode="w",
    ) as archive:
        payload = SELECTED_SOURCE.encode("utf-8")
        info = tarfile.TarInfo("main.py")
        info.size = len(payload)
        info.mode = 0o644
        info.mtime = 0
        info.uid = info.gid = 0
        info.uname = info.gname = ""
        archive.addfile(
            info,
            io.BytesIO(payload),
        )

ARCHIVE_PATH.write_bytes(raw.getvalue())

with tarfile.open(
    ARCHIVE_PATH,
    "r:gz",
) as archive:
    assert archive.getnames() == ["main.py"]
    archived = (
        archive.extractfile("main.py")
        .read()
        .decode("utf-8")
    )

assert archived == SELECTED_SOURCE

print({
    "selected": SELECTED_NAME,
    "main_path": str(MAIN_PATH),
    "submission": str(ARCHIVE_PATH),
    "submission_bytes": (
        ARCHIVE_PATH.stat().st_size
    ),
    "main_sha256": hashlib.sha256(
        SELECTED_SOURCE.encode("utf-8")
    ).hexdigest(),
    "archive_sha256": hashlib.sha256(
        ARCHIVE_PATH.read_bytes()
    ).hexdigest(),
})
```

```python
RUN_EXTENDED = False

if RUN_EXTENDED and EVAL_AVAILABLE:
    EXTENDED_SEEDS = (
        27203,
        27221,
        27241,
    )
    extended_rows = []

    for opponent in (
        ANCHOR_NAME,
        *CONTROL_NAMES,
    ):
        for seed in EXTENDED_SEEDS:
            extended_rows.extend(
                symmetric_pair(
                    SELECTED_NAME,
                    opponent,
                    seed,
                )
            )

    summaries = []
    for opponent in sorted({
        row["opponent"]
        for row in extended_rows
    }):
        group = [
            row
            for row in extended_rows
            if row["opponent"] == opponent
        ]
        summaries.append({
            "opponent": opponent,
            "games": len(group),
            "wins": sum(
                row["result"] == "win"
                for row in group
            ),
            "mean_money": round(
                statistics.mean(
                    row["candidate_money"]
                    for row in group
                ),
                2,
            ),
            "minimum_money": min(
                row["candidate_money"]
                for row in group
            ),
            "mean_margin": round(
                statistics.mean(
                    row["margin"]
                    for row in group
                ),
                2,
            ),
            "mean_leftover_value": round(
                statistics.mean(
                    row["candidate_leftover"]
                    for row in group
                ),
                2,
            ),
        })

    print(json.dumps(summaries, indent=2))
```
